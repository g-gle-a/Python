## Agentic AI System (Full Architecture + Code)
We’ll build a 3-agent system using LangChain.
Architecture
User Request
     ↓
[Research Agent] → summarizes literature
     ↓
[Code Agent] → runs RDKit experiments
     ↓
[Writing Agent] → generates paper/report

```python
rom langchain.chat_models import ChatOpenAI
from langchain.agents import initialize_agent, Tool
import subprocess

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

# --- Tool 1: Research ---
def research_tool(query):
    prompt = f"Summarize key techniques in {query}"
    return llm.predict(prompt)

# --- Tool 2: Run Code ---
def run_code_tool(code_str):
    with open("temp_script.py", "w") as f:
        f.write(code_str)
    result = subprocess.run(["python", "temp_script.py"], capture_output=True, text=True)
    return result.stdout

# --- Tool 3: Writing ---
def writing_tool(content):
    prompt = f"Write a scientific report based on:\n{content}"
    return llm.predict(prompt)

tools = [
    Tool(name="Research", func=research_tool, description="Summarize literature"),
    Tool(name="CodeRunner", func=run_code_tool, description="Execute Python code"),
    Tool(name="Writer", func=writing_tool, description="Generate paper drafts"),
]

agent = initialize_agent(tools, llm, agent="zero-shot-react-description", verbose=True)

response = agent.run("Analyze molecular similarity, run experiments, and write a report")
print(response)
```
